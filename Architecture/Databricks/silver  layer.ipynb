### Reading calender data


```python
df_calendar = spark.read.format("csv").option("header", True).option("inferSchema", True).load(
    "abfss://bronze@srinadhgen2007.dfs.core.windows.net/Calendar")
```

```python
df_customers = spark.read.format("csv").option("header", True).option("inferSchema", True).load("abfss://bronze@srinadhgen2007.dfs.core.windows.net/Customers")
```

```python
df_product_categories =spark.read.format("csv").option("header", True).option("inferSchema", True).load(
    "abfss://bronze@srinadhgen2007.dfs.core.windows.net/Product_Categories")
```

```python
df_products = spark.read.format("csv").option("header", True).option("inferSchema", True).load(
    "abfss://bronze@srinadhgen2007.dfs.core.windows.net/Products")
```

```python
df_returns = spark.read.format("csv").option("header", True).option("inferSchema", True).load(
    "abfss://bronze@srinadhgen2007.dfs.core.windows.net/Returns")
```

```python
paths = [
    "abfss://bronze@srinadhgen2007.dfs.core.windows.net/Sales_2015/sales_2015.csv",
    "abfss://bronze@srinadhgen2007.dfs.core.windows.net/Sales_2016/sales_2016.csv",
    "abfss://bronze@srinadhgen2007.dfs.core.windows.net/Sales_2017/sales_2017.csv"
]

df_sales = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(paths)
)

display(df_sales)
```

```python
df_territories = spark.read.format("csv").option("header", True).option("inferSchema", True).load(
    "abfss://bronze@srinadhgen2007.dfs.core.windows.net/Territories")
```

```python
df_product_sub_categories = spark.read.format("csv").option("header", True).option("inferSchema", True).load(
    "abfss://bronze@srinadhgen2007.dfs.core.windows.net/Product_Subcategories")
```

#Transformations


```python
display(df_calendar)
```

```python
df_calendar = df_calendar.withColumn("Month",month(col("Date"))).withColumn("Year",year(col("Date")))
```

```python
display(df_calendar)
```

```python
df_calendar.write.format('parquet').mode('append').option('path', "abfss://silver@srinadhgen2007.dfs.core.windows.net/calendar").save()
```

```python
df_customers= df_customers.withColumn('fullname',concat_ws(" ",col("Prefix"),col("FirstName"),col("LastName")))
```

```python
df_customers.write.format('parquet').mode('append').option('path', "abfss://silver@srinadhgen2007.dfs.core.windows.net/customers").save()
```

```python
display(df_product_sub_categories)
```

```python
df_product_categories.write.format('parquet').mode('append').option('path', "abfss://silver@srinadhgen2007.dfs.core.windows.net/Product_Categories").save()
```
